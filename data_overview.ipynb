# Description: This notebook serves as the initial analysis of the data. It will create a table share useful information about the frequency of each cell type in each sample.

```python
# Importing libraries

import pandas as pd
import sqlite3
from pathlib import Path
```

```python
# connecting the database and creating the object 
DB_NAME = "cell_count.db"
conn = sqlite3.connect(DB_NAME)
```

```python
# Loading the cell counts table as it has the relevant information 

cell_counts = pd.read_sql(
    """
    SELECT *
    FROM cell_counts
    """,
    conn
)

# cell_counts.head()
```

```python
# calculating the total cells per sample

total_counts = (cell_counts.groupby("sample_id")["count"].sum().reset_index())
total_counts.rename(columns={"count": "total_count"},inplace=True)
# total_counts.head()
```

```python
#  creating the summary table 

summary = cell_counts.merge( total_counts, on="sample_id")
summary["percentage"] = (summary["count"]/ summary["total_count"]* 100)
# summary.head()
```

```python
#  restructuring the order of the columns

summary = summary[
    [
        "sample_id",
        "total_count",
        "population",
        "count",
        "percentage"
    ]
]

# adjusting formatting of table
summary["count"] = summary["count"].astype(int)
summary["percentage"] = summary["percentage"].round(2)
summary["total_count"] = summary["total_count"].astype(int)

# verifying analysis
print("Rows:", len(summary))
print("Unique samples:", summary["sample_id"].nunique())
print("Unique populations:", summary["population"].nunique())
# summary.head(10)


```

```python
# saving output table in the output folder

Path("outputs").mkdir(exist_ok=True)
summary.to_csv(
    "outputs/sample_population_frequencies.csv",
    index=False
)

print("Summary table saved.")
```
